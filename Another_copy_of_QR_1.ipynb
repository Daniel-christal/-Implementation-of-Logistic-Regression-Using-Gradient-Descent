```python
import torch
import torch.nn as nn  # Neural network module
import matplotlib.pyplot as plt  # For plotting
```

```python
torch.manual_seed(71)
x = torch.randn(70, 1) # Define x
e = torch.randint(-8, 9, (70, 1), dtype=torch.int) # Corrected syntax for randint
y = 2 * x + 1 + e
print(y.shape)
```

```python
# Plot the original data
plt.scatter(x, y, color='red') # Changed X to x
plt.xlabel('x')
plt.ylabel('y')
plt.title('Generated Data for Linear Regression')
plt.show()
```

```python
# Define the Linear Model Class
class Model(nn.Module):
    def __init__(self, in_features, out_features):
        super().__init__()
        self.linear = nn.Linear(in_features, out_features) # Corrected nn.linear to nn.Linear

    def forward(self, x):
        y_pred = self.linear(x)
        return y_pred # Added return statement for forward method
```

```python
# Initialize the Model
torch.manual_seed(59)  # Ensure same initial weights
model = Model(1, 1)
```

```python
# Print Initial Weights and Bias
initial_weight = model.linear.weight.item()
initial_bias = model.linear.bias.item()
print("\nName: Daniel C")
print("Register No: 212223240023")
print(f'Initial Weight: {initial_weight:.8f}, Initial Bias: {initial_bias:.8f}\n')
```

```python
# Define Loss Function & Optimizer
loss_function = nn.MSELoss() # Corrected nn.MSE loss() to nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.0001) # Corrected ln to lr
```

```python
# Train the Model
epochs = 100
losses = []

for epoch in range(1, epochs + 1):  # Loop over epochs
    optimizer.zero_grad()
    y_pred = model(x)
    loss = loss_function(y_pred, y) # Corrected loss_function(y_pred_y) to loss_function(y_pred, y)
    losses.append(loss.item())
    loss.backward()
    optimizer.step()




    # Print loss, weight, and bias for EVERY epoch
    print(f'epoch: {epoch:2}  loss: {loss.item():10.8f}  '
          f'weight: {model.linear.weight.item():10.8f}  '
          f'bias: {model.linear.bias.item():10.8f}')
```

```python
# Plot Loss Curve
plt.plot(range(epochs), losses, color='blue')
plt.ylabel('Loss')
plt.xlabel('Epoch')
plt.title('Loss Curve')
plt.show()
```

```python
# Final Weights & Bias
final_weight = model.linear.weight.item()
final_bias = model.linear.bias.item()
print("\nName: Daniel C")
print("Register No: 2122232440023")
print(f'\nFinal Weight: {final_weight:.8f}, Final Bias: {final_bias:.8f}')
```

```python
plt.plot(range(epochs),losses)
plt.ylabel('Loss')
plt.xlabel('epoch')
plt.show()
x1 = torch.tensor([x.min().item(), x.max().item()]) # Changed X to x
y1 = x1 * final_weight + final_bias # Compute corresponding y-values using trained model
```

```python
# Plot Original Data & Best-Fit Line
plt.scatter(x, y, label="Original Data") # Changed X to x
plt.plot(x1, y1, 'r', label="Best-Fit Line")
plt.xlabel('x')
plt.ylabel('y')
plt.title('Trained Model: Best-Fit Line')
plt.legend()
plt.show()
```

```python
# Prediction for x = 120
x_new = torch.tensor([[120.0]])  # New input as a tensor
y_new_pred = model(x_new).item()  # Predict using trained model
print("\nName: Daniel C")
print("Register No: 212223240023")
print(f"\nPrediction for x = 120: {y_new_pred:.8f}")
```
